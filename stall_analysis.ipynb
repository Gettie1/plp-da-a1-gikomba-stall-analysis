```python
# Daily sales for 14 consecutive days (in KES)

daily_sales = [
    2300, 1950, 2800, 2100, 3600, 5200, 4100,
    2500, 2200, 3100, 2400, 3900, 5600, 4700
]

# Corresponding days of the week

days = [
    "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun",
    "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"
]

print("Daily sales data loaded successfully!")
print("Number of sales records:", len(daily_sales))
print("Number of days:", len(days))
```

```python
# Calculate total sales
total_sales = sum(daily_sales)

# Calculate average daily sales
average_sales = total_sales / len(daily_sales)

# Find the highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the days with the highest and lowest sales
highest_day = days[daily_sales.index(highest_sales)]
lowest_day = days[daily_sales.index(lowest_sales)]

# Display the results
print("SALES PERFORMANCE SUMMARY")
print("-------------------------")
print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,.2f} ({highest_day})")
print(f"Lowest Sales: KES {lowest_sales:,.2f} ({lowest_day})")
```

```python
def week_totals(sales):
    # Calculate sales for the first week
    first_week = sum(sales[:7])

    # Calculate sales for the second week
    second_week = sum(sales[7:14])

    # Return both weekly totals
    return first_week, second_week


# Call the function
week1, week2 = week_totals(daily_sales)

print("WEEKLY SALES SUMMARY")
print("--------------------")
print(f"Week 1: KES {week1:,.2f}")
print(f"Week 2: KES {week2:,.2f}")
```

```python
# Calculate the difference between the two weeks
sales_difference = week2 - week1

# Calculate percentage growth
growth_percentage = (sales_difference / week1) * 100

print("WEEKLY GROWTH ANALYSIS")
print("----------------------")
print(f"Sales difference: KES {sales_difference:,.2f}")
print(f"Weekly growth: {growth_percentage:.2f}%")
```

```python
def classify_day(amount, average):
    """Classify daily sales based on the overall average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"
```

```python
# Create an empty dictionary to store classifications
# Create a dictionary to count each category
classification_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

average_sales = sum(daily_sales) / len(daily_sales)
# Classify each day's sales
for i in range(len(daily_sales)):
    category = classify_day(daily_sales[i], average_sales)

    # Increase the appropriate category count
    classification_counts[category] += 1

# Display the results
print("Sales Classification Results")

for category, count in classification_counts.items():
    print(f"{category}: {count} days")
```

```python
# Create empty lists
weekday_sales = []
weekend_sales = []

# Separate sales into weekdays and weekends
for i in range(len(daily_sales)):

    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])

    else:
        weekend_sales.append(daily_sales[i])

# Display the two groups
print("Weekday sales:", weekday_sales)
print("Weekend sales:", weekend_sales)
```

```python
# Calculate total sales
total_weekday_sales = sum(weekday_sales)
total_weekend_sales = sum(weekend_sales)

# Calculate average sales
average_weekday_sales = total_weekday_sales / len(weekday_sales)
average_weekend_sales = total_weekend_sales / len(weekend_sales)

# Display results
print("WEEKDAY VS WEEKEND SALES")
print("------------------------")

print(f"Total weekday sales: KES {total_weekday_sales:,.2f}")
print(f"Total weekend sales: KES {total_weekend_sales:,.2f}")

print(f"Average weekday sales: KES {average_weekday_sales:,.2f}")
print(f"Average weekend sales: KES {average_weekend_sales:,.2f}")
```

```python
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}
```

```python
def check_reorder(stock_data):
    """Return products whose stock is at or below reorder level."""
    products_to_reorder = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            products_to_reorder.append(product)

    return products_to_reorder
```

```python
def calculate_profit_margin(cost_price, selling_price):

    profit = selling_price - cost_price

    profit_margin = (profit / selling_price) * 100

    return profit_margin
```

```python
print("GROSS PROFIT MARGINS")
print("--------------------------------")

for product, details in stock.items():

    cost = details["cost"]
    selling = details["price"]

    margin = calculate_profit_margin(cost, selling)

    print(f"{product}: {margin:.2f}%")
```

```python

# TASK 6: STALL PERFORMANCE REPORT

print("=" * 55)
print("              STALL PERFORMANCE REPORT")
print("                   GIKOMBA MARKET")
print("=" * 55)

# 1. SALES PERFORMANCE
print("\n1. SALES PERFORMANCE")
print("-" * 55)

print(f"Total Sales:          KES {total_sales:,.2f}")
print(f"Average Daily Sales:  KES {average_sales:,.2f}")
print(f"Highest Daily Sales:  KES {highest_sales:,.2f}")
print(f"Lowest Daily Sales:   KES {lowest_sales:,.2f}")

# 2. WEEKLY PERFORMANCE
print("\n2. WEEKLY PERFORMANCE")
print("-" * 55)

print(f"Week 1 Sales:         KES {week1:,.2f}")
print(f"Week 2 Sales:         KES {week2:,.2f}")
print(f"Weekly Growth:        {growth_percentage:.2f}%")

# 3. SALES CLASSIFICATION
print("\n3. SALES CLASSIFICATION")
print("-" * 55)

for category, count in classification_counts.items():
    print(f"{category} Days: {count}")

# 4. INVENTORY STATUS
print("\n4. INVENTORY STATUS")
print("-" * 55)

for product, details in stock.items():

    quantity = details["qty"]
    reorder_level = details["reorder"]

    if quantity <= reorder_level:
        status = "REORDER REQUIRED"
    else:
        status = "Stock Sufficient"

    print(f"{product}: {quantity} kg - {status}")

# 5. PRODUCT GROSS PROFIT MARGINS
print("\n5. PRODUCT GROSS PROFIT MARGINS")
print("-" * 55)

for product, details in stock.items():

    margin = calculate_profit_margin(
        details["cost"],
        details["price"]
    )

    print(f"{product}: {margin:.2f}%")

print("\n" + "=" * 55)
print("                 END OF REPORT")
print("=" * 55)
```
